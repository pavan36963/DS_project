## Program 1: Linear Regression - House Price Prediction

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# House sizes in square feet
X = np.array([[1000], [1200], [1500], [1800], [2000]])

# House prices in lakhs
y = np.array([50, 55, 65, 80, 90])

# Linear regression model
model = LinearRegression()
model.fit(X, y)

# Predict price for 1700 sq.ft house
new_house = np.array([[1700]])
predicted_price = model.predict(new_house)

print("----- Linear Regression -----")
print("Training House Sizes:")
print(X.flatten())
print("\nTraining House Prices:")
print(y)
print("\nPredicted Price for 1700 sq.ft house:")
print(predicted_price[0])
```

## Program 2: Logistic Regression - House Purchase Classification

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# House prices in lakhs
X = np.array([[40], [50], [60], [70], [80], [90]])

# 0 = Not Purchased, 1 = Purchased
y = np.array([0, 0, 0, 1, 1, 1])

# Logistic regression model
model = LogisticRegression()
model.fit(X, y)

# Predict whether a customer will purchase a house priced at 75 lakhs
test_price = np.array([[75]])
prediction = model.predict(test_price)

# Calculate probabilities
probability = model.predict_proba(test_price)

print("----- Logistic Regression -----")
print("Training House Prices:")
print(X.flatten())
print("\nPurchase Labels:")
print(y)
print("\nTest House Price:", test_price[0][0])
print("\nPurchase Prediction:")
print(prediction[0])
print("\nPrediction Probabilities:")
print(probability)
```
