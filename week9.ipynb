## Program 1: Bernoulli Distribution

```python
import numpy as np
from scipy.stats import bernoulli

# Probability of passing an exam
p = 0.7

# Generate outcomes for 10 students: 1 = Pass, 0 = Fail
bernoulli_samples = bernoulli.rvs(p, size=10)

print("----- Bernoulli Distribution -----")
print("Probability of Passing:", p)
print("Student Outcomes (1=Pass, 0=Fail):")
print(bernoulli_samples)

# Probability of passing
print("\nProbability of Pass:")
print(bernoulli.pmf(1, p))

# Probability of failing
print("Probability of Fail:")
print(bernoulli.pmf(0, p))
```

## Program 2: Binomial Distribution

```python
from scipy.stats import binom

# Number of students and probability of passing
n = 10
p = 0.7

# Probability that exactly 8 students pass
k = 8
binomial_probability = binom.pmf(k, n, p)

print("----- Binomial Distribution -----")
print("Number of Students:", n)
print("Probability of Passing:", p)
print(f"Probability that exactly {k} students pass:")
print(binomial_probability)

# Generate results for 5 random experiments
random_binomial = binom.rvs(n, p, size=5)
print("\nRandom Binomial Outcomes:")
print(random_binomial)
```

## Program 3: Z-Test

```python
import numpy as np
from statsmodels.stats.weightstats import ztest

# Sample marks of students
sample_marks = np.array([78, 82, 85, 88, 90, 76, 95, 89, 84, 91])

# Population mean to compare against
population_mean = 80

# H0: mean marks equal 80
# H1: mean marks are different from 80
z_stat, p_value = ztest(sample_marks, value=population_mean)

print("----- Z-Test -----")
print("Sample Marks:")
print(sample_marks)
print("\nPopulation Mean:", population_mean)
print("Z-Statistic:", round(z_stat, 3))
print("P-Value:", round(p_value, 4))

# Decision using 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Result: Reject Null Hypothesis")
else:
    print("Result: Fail to Reject Null Hypothesis")
```

## Program 4: Independent t-Test

```python
import numpy as np
from scipy import stats

# Marks of two different classes
class_A = np.array([78, 82, 85, 88, 90])
class_B = np.array([72, 75, 70, 68, 74])

# H0: both class means are equal
# H1: class means are different
t_stat, p_value = stats.ttest_ind(class_A, class_B)

print("----- Independent t-Test -----")
print("Class A Marks:")
print(class_A)
print("\nClass B Marks:")
print(class_B)
print("\nt-Statistic:", round(t_stat, 2))
print("P-Value:", round(p_value, 4))

# Decision using 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Result: Reject Null Hypothesis")
else:
    print("Result: Fail to Reject Null Hypothesis")
```
