```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df['species']=iris.target
plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'],bins=20,kde=True)
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.show()
# boxplot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['sepal length (cm)'])
plt.title('Boxplot of Sepal Length')
plt.show()
#pie chart

species_counts=df['species'].value_counts()
plt.figure(figsize=(6,4))
plt.pie(species_counts,labels=iris.target_names,autopct='%1.1f%%')
plt.title('Distribution of Species(pie chart)')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df['species']=iris.target
plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'],bins=20,kde=True)
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')

species_counts=df['species'].value_counts()
plt.figure(figsize=(6,4))
plt.pie(species_counts,labels=iris.target_names,autopct='%1.1f%%')
plt.title('pie chart')
plt.show()
```

```python
from matplotlib import figure
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'],y=df['sepal width (cm)'],hue=df['species'])
plt.title('Scatterplot: Sepal Length vs sepal Width')
plt.show()
plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title('Line chart of Sepal Length')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length')
plt.show()
plt.figure(figsize=(6,4))
sns.barplot(x=df['species'],y=df['sepal length (cm)'])
plt.title('Bar chart of Sepal Length by Species')
plt.xticks([0,1,2],iris.target_names)
plt.xlabel('Species')
plt.ylabel('Sepal Length')
plt.show()
```

```python
#heatmap
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(),annot=True,cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
#bubble chart
plt.figure(figsize=(6,4))
plt.scatter(df['sepal length (cm)'],df['sepal width (cm)'],s=df['petal length (cm)']*20,alpha=0.5,c=df['species'])
plt.title('Bubble Chart: Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
#plair plot
sns.pairplot(df.iloc[:,:4])
plt.suptitle("Pair Plot of Iris Features",y=1.02)
plt.show()
```

```python
#violin Plot
plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'],y=df['sepal length (cm)'])
plt.title('Violin Plot of Sepal Length by Species')
plt.xticks([0,1,2],iris.target_names)
plt.xlabel('Species')
plt.ylabel('Sepal Length')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')

plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'],bins=20,kde=True)
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')
plt.show()

# boxplot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['total_bill'])
plt.title('Boxplot of Total Bill')
plt.show()

# pie chart for 'sex'
sex_counts = df['sex'].value_counts()
plt.figure(figsize=(6,4))
plt.pie(sex_counts,labels=sex_counts.index,autopct='%1.1f%%')
plt.title('Distribution of Diners by Sex')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['total_bill'],y=df['tip'],hue=df['day'])
plt.title('Scatterplot: Total Bill vs Tip by Day')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
plt.figure(figsize=(6,4))
plt.plot(df['total_bill'])
plt.title('Line chart of Total Bill')
plt.xlabel('Sample Index')
plt.ylabel('Total Bill ($)')
plt.show()
plt.figure(figsize=(6,4))
sns.barplot(x=df['day'],y=df['total_bill'])
plt.title('Bar chart of Total Bill by Day')
plt.xlabel('Day of Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

```python
#heatmap
plt.figure(figsize=(6,4))
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Tips Dataset Numerical Features')
plt.show()

#bubble chart
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['total_bill'], y=df['tip'], size=df['size'], hue=df['time'], sizes=(20, 400), alpha=0.7)
plt.title('Bubble Chart: Total Bill vs Tip by Time of Day (Size indicates Party Size)')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()

#pair plot
sns.pairplot(df[['total_bill', 'tip', 'size', 'day']], hue='day')
plt.suptitle("Pair Plot of Tips Dataset Features by Day", y=1.02)
plt.show()
```

```python
#violin Plot
plt.figure(figsize=(6,4))
sns.violinplot(x=df['day'],y=df['total_bill'])
plt.title('Violin Plot of Total Bill by Day')
plt.xlabel('Day of Week')
plt.ylabel('Total Bill ($)')
plt.show()
```
